## Part 1 - Lets-Plot Essentials

Create a scatterplot of `Sepal.Length` vs. `Sepal.Width`

```python
ggplot(data=iris) + geom_point(
      mapping=aes(x='Sepal.Length', y='Sepal.Width')
  )
```

Create a scatterplot of `Sepal.Length` vs. `Sepal.Width` but plot points in blue

```python
ggplot(data=iris) + geom_point(
      mapping=aes(x='Sepal.Length', y='Sepal.Width'), color = 'blue'
  )
```

Create a scatterplot of `Sepal.Length` vs. `Sepal.Width` with different shapes for each `Species`

```python
ggplot(data=iris) + geom_point(
      mapping=aes(x='Sepal.Length', y='Sepal.Width', shape='Species')
)
```

Create a histogram of `Sepal.Length` using `geom_histogram`


```python
ggplot(data=iris) + geom_histogram(
      mapping=aes(x='Sepal.Length')
  )
```

Change the width of bins in the histogram of `Sepal.Length` to any value you want using `binwidth`

```python
ggplot(data=iris) + geom_histogram(
      mapping=aes(x='Sepal.Length'), binwidth=0.5
  )
```

## Part 2 - Facets

Create a separate scatterplot of `Petal.Length` vs `Petal.Width` for each `Species` of iris.

```python
ggplot(data=iris) + geom_point(
      mapping=aes(x='Petal.Length', y='Petal.Width', shape='Species')
) + facet_wrap('Species')
```

Make the scales of the x-axis not match each other

```python
ggplot(data=iris) + geom_point(
      mapping=aes(x='Petal.Length', y='Petal.Width', shape='Species')
) + facet_wrap('Species', scales='free_x')
```

## Part 3 - Customizing your Plot

Create a scatterplot with the following specifications:
1.   The x-axis is `Sepal.Length` and y-axis is `Sepal.Width`
2.   The shapes AND colors of each point should correspond to a different `Species`
3.   Add a "jitter" to the position of points to compensate for overlaps
4.   Label the x-axis "Length" and y-axis "Width" appropriately and the title "Sample Sepal Measurements in Irises"
5.   Set the coordinate limits to be between 4 and 8.5 for the x-axis and between 1.5 and 5 for the y-axis (using `coord_cartesian`)
6.  Choose any theme that is not the default. Type `help('lets_plot.plot.theme_set')` in a cell to see a list of themes!
7.   Adjust the title so that it is centered with `+ theme(plot_title=element_text(hjust=0.5))`

```python
myplot = ggplot(data=iris) + geom_point(
      mapping=aes(x='Sepal.Length', y='Sepal.Width', shape='Species', color='Species',),
      position='jitter'
) + labs(
      x='Length', y='Width', title='Sample Sepal Measurements in Irises'
) + coord_cartesian(xlim=[4, 8.5], ylim=[1.5, 5]
) + theme(plot_title=element_text(hjust=0.5)) + theme_bw() + flavor_darcula()
```
